Missing Values 

```python
data = {
    "Name": ["Arun", "Atul", "Pratik", "Rahul", "Sneha", "Priya", "Rohan", "Aisha"],
    "Age": [25, 30, np.nan, 28, 32, np.nan, 26, 29],
    "Salary": [30000, np.nan, 45000, 52000, np.nan, 48000, 35000, 60000],
    "City": ["Pune", "Mumbai", "Delhi", np.nan, "Pune", "Mumbai", np.nan, "Delhi"],
    "Experience": [2, 5, np.nan, 4, 7, 3, np.nan, 6]
}

df = pd.DataFrame(data)

df
```

Checking Null Values 

```python
df.isnull()
```

Total number of Null Values

```python
df.isnull().sum()
```

TO CHECK VALUES WHICH ARE NOT NULL

```python
df[df.notnull().all(axis=1)]
```

To drop null Values rows 

```python
df.dropna()
```

Drop only those rows who have less 3 non values

```python
df.dropna(thresh=3)
```

Filling out Missing Values

```python
df.fillna(0)
```

```python
df = df.fillna(df.mean(numeric_only=True))
```

```python
df
```

```python
df["City"]= df["City"].fillna("Agra")
```
