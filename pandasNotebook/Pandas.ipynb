Merging  Two Dataframes

```python
df1 = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Aarav", "Priya", "Rahul", "Sneha", "Vikram"],
    "Department_ID": [1, 2, 1, 3, 2]
})

df1
```

```python
df2 = pd.DataFrame({
    "Department_ID": [1, 2, 3, 4],
    "Department": ["IT", "HR", "Finance", "Marketing"],
    "Location": ["Pune", "Mumbai", "Delhi", "Bangalore"]
})

df2
```

```python
pd.merge(df1, df2, on="Department_ID", how="inner")
```

```python
pd.merge(df1, df2, on="Department_ID", how="outer")
```

```python


data = {
    "Ticket_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112],
    "Passenger": ["Rahul", "Priya", "Amit", "Sneha", "Karan", "Anjali",
                  "Rohit", "Neha", "Vikram", "Pooja", "Arjun", "Meera"],
    "Train": ["Rajdhani", "Shatabdi", "Rajdhani", "Duronto",
              "Shatabdi", "Rajdhani", "Duronto", "Shatabdi",
              "Rajdhani", "Duronto", "Shatabdi", "Rajdhani"],
    "From": ["Pune", "Mumbai", "Pune", "Delhi",
             "Mumbai", "Pune", "Delhi", "Mumbai",
             "Pune", "Delhi", "Mumbai", "Pune"],
    "To": ["Delhi", "Pune", "Delhi", "Mumbai",
           "Pune", "Delhi", "Mumbai", "Pune",
           "Delhi", "Mumbai", "Pune", "Delhi"],
    "Class": ["AC 2-Tier", "AC Chair", "AC 3-Tier", "Sleeper",
              "AC Chair", "AC 2-Tier", "Sleeper", "AC Chair",
              "AC 3-Tier", "Sleeper", "AC Chair", "AC 2-Tier"],
    "Passengers": [2, 1, 3, 2, 1, 4, 2, 3, 1, 2, 4, 2],
    "Fare": [2400, 1200, 3600, 1600, 1200, 4800,
             1600, 3600, 1200, 1600, 4800, 2400]
}

df = pd.DataFrame(data)

df
```

```python
df.groupby("Train")["Fare"]
```

```python
fare = df.groupby("Train")["Fare"]
for x,v in fare:
    print("\n",x,"\n",v)
```
