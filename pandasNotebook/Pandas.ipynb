Missing Values 

```python
data = {
    "Name": ["Arun", "Atul", "Pratik", "Rahul", "Sneha", "Priya", "Rohan", "Aisha"],
    "Age": [25, 30, np.nan, 28, 32, np.nan, 26, 29],
    "Salary": [30000, np.nan, 45000, 52000, np.nan, 48000, 35000, 60000],
    "City": ["Pune", "Mumbai", "Delhi", np.nan, "Pune", "Mumbai", np.nan, "Delhi"],
    "Experience": [2, 5, np.nan, 4, 7, 3, np.nan, 6]
}

df = pd.DataFrame(data)

df
```

Merging  Two Dataframes

```python
df1 = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Aarav", "Priya", "Rahul", "Sneha", "Vikram"],
    "Department_ID": [1, 2, 1, 3, 2]
})

df1
```

```python
df2 = pd.DataFrame({
    "Department_ID": [1, 2, 3, 4],
    "Department": ["IT", "HR", "Finance", "Marketing"],
    "Location": ["Pune", "Mumbai", "Delhi", "Bangalore"]
})

df2
```

```python
pd.merge(df1, df2, on="Department_ID", how="inner")
```

```python
pd.merge(df1, df2, on="Department_ID", how="outer")
```
