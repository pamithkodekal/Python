#Series

A Pandas Series is a one-dimensional, labeled array capable of holding data of any type

```python
!pip install numpy pandas
import numpy as np
import pandas as pd
```

```python
strings = ["Alpha","Beta","Gama"]
integers = [10,20,30]
array = np.array([40,50,60])
dictionary = {1:"JAVA" , 2:"PYTHON" , 3:"C++"}
```

#Priting String

```python
print(pd.Series(strings))
```

#print integers

```python
print(pd.Series(integers))
```

#Print array

```python
print(pd.Series(array))
```

#Print dictionary

```python
print(pd.Series(dictionary))
```

#Replacing Index

```python
print(pd.Series(integers,index = ["a","b","c"]) )#Single inverted commas can be  used too example'a','b'
```

#Using diff series for index

```python
print(pd.Series(integers,index = strings))
```

```python
data = {
    "Name": ["Aarav", "Priya", "Rahul"],
    "Salary": [45000, 52000, 48000],
    "City": ["Pune", "Mumbai", "Delhi"]
}

df = pd.DataFrame(data)

df
```

```python
data = [
    ["Aarav", 45000, "Pune"],
    ["Priya", 52000, "Mumbai"],
    ["Rahul", 48000, "Delhi"]
]

df = pd.DataFrame(data, columns=["Name", "Salary", "City"])

df
```
